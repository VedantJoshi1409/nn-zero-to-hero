```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
words = open("names.txt", "r").read().splitlines()
words
```

```python
def toIdx(c): 
    if c == '.': return 26
    return ord(c)-ord('a')

# freqs = [[0]*28 for _ in range(28)]
freqs = torch.zeros((27,27), dtype=torch.int32)
for word in words:
    word = ['.'] + list(word) + ['.']
    for i in range(len(word)-1):
        freqs[toIdx(word[i])][toIdx(word[i+1])]+=1
freqs
```

```python
g = torch.Generator().manual_seed(2147483647)
p = freqs.float()
p /= p.sum(1, keepdim=True)
for i in range(10):
    idx = 26
    name = ""
    while True:
        idx = torch.multinomial(p[idx], num_samples=1, replacement=True, generator=g).item()
        if idx == 26: break
        name += chr(ord('a')+idx)
    print(name)
```

```python
log_likelihood = 0
n = 0
xs = []
ys = []
for word in words:
    word = ['.'] + list(word) + ['.']
    bigrams = [(f,s) for f, s in zip(word, word[1:])]
    for b in bigrams:
        log_likelihood += torch.log(p[toIdx(b[0])][toIdx(b[1])])
        n+=1
        xs.append(toIdx(b[0]))
        ys.append(toIdx(b[1]))
nll = -log_likelihood/n
nll
xs = torch.tensor(xs)
ys = torch.tensor(ys)
nll
```

```python
import torch.nn.functional as F
xenc = F.one_hot(xs, num_classes=27).float()
W = torch.randn((27, 27), requires_grad=True)
```

```python
for i in range(1000):
    logits = xenc @ W
    counts = logits.exp()
    probs = counts / counts.sum(1,keepdim=True)
    loss = -probs[torch.arange(ys.shape[0]), ys].log().mean()
    print(loss)
    W.grad = None
    loss.backward()
    W.data -= 1 * W.grad
```

```python
g = torch.Generator().manual_seed(2147483647)
p = W.data.exp()
p /= p.sum(1, keepdim=True)
for i in range(10):
    idx = 26
    name = ""
    while True:
        idx = torch.multinomial(p[idx], num_samples=1, replacement=True, generator=g).item()
        if idx == 26: break
        name += chr(ord('a')+idx)
    print(name)
```

```python
n = 0
xs = []
ys = []
for word in words:
    word = ['.'] + list(word) + ['.']
    for i in range(2, len(word)):
        xs.append(toIdx(word[i-2])*27 + toIdx(word[i-1]))
        ys.append(toIdx(word[i]))
xs = torch.tensor(xs)
ys = torch.tensor(ys)
```

```python
W = torch.randn((27*27, 27), requires_grad=True)
```

```python
# we have input shape 1x(27*27) 
xenc = F.one_hot(xs, num_classes=27*27).float()
for i in range(100):
    logits = xenc @ W
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)
    loss = -probs[torch.arange(ys.shape[0]), ys].log().mean()
    W.grad = None
    loss.backward()
    W.data -= 50 * W.grad

    print(loss.data)
```

```python
probs = W.data.exp()
probs /= probs.sum(1, keepdim=True)

for i in range(10):
    fst = 26
    snd = i
    name = chr(ord('a')+snd)
    while True:
        tmp = torch.multinomial(probs[fst*27+snd], num_samples=1, replacement=True, generator=g).item()
        fst = snd
        snd = tmp
        if snd == 26: break
        name += chr(ord('a')+snd)
    print(name)
```
